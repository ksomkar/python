```python
Write a program to check the given num is prime or not 
```

```python

n = int(input("Enter a number: "))

if n <= 1:
    print("Not Prime")
else:
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
```
